## 1. Eurostat API: Criminalidad y Datos Socioeconómicos

Utilizaremos la API de Eurostat para obtener datos sobre delitos registrados.
Dataset de ejemplo: `crim_off_cat` (Recorded offences by offence category - police data).

```python
import eurostat

# Código del dataset para crímenes
code = 'crim_off_cat'

# Filtro para obtener datos de un país específico (ej. España 'ES', Alemania 'DE') y un tipo de crimen
# 'ICCS' es la clasificación internacional de crímenes. 'ICCS0101' es Homicidio intencional.
filter_pars = {'geo': ['ES', 'DE', 'FR', 'IT'], 'iccs': 'ICCS0101', 'unit': 'NR'}

# Obtener datos
df = eurostat.get_data_df(code)

# Filtrar el dataframe localmente si la librería no soporta filtrado en la llamada (depende de la versión)
# La librería eurostat devuelve todo el dataset a veces, así que filtramos con pandas
df_filtered = df[df['geo\TIME_PERIOD'].isin(['ES', 'DE', 'FR', 'IT']) & (df['iccs'] == 'ICCS0101')]

# Limpiar y transponer para graficar
# Las columnas de años suelen ser strings tipo '2020', '2021', etc.
year_cols = [col for col in df.columns if col.isdigit()]
df_plot = df_filtered[['geo\TIME_PERIOD'] + year_cols].set_index('geo\TIME_PERIOD').T
df_plot.index = df_plot.index.astype(int)

# Graficar
plt.figure(figsize=(10, 6))
df_plot.plot(ax=plt.gca())
plt.title('Homicidios Intencionales (ICCS0101) - Comparativa')
plt.xlabel('Año')
plt.ylabel('Número de casos')
plt.legend(title='País')
plt.show()

print("Muestra de datos:")
print(df_filtered.head())
```

## 2. OSM Overpass API: Estructura Urbana

Consultaremos la API de Overpass para obtener elementos de infraestructura urbana.
Ejemplo: Comisarías de policía (`amenity=police`) en un área específica (ej. Madrid).

```python
import requests
import folium
import time
import os

# Lista de servidores Overpass para intentar (Failover)
overpass_urls = [
    "https://overpass-api.de/api/interpreter",
    "https://maps.mail.ru/osm/tools/overpass/api/interpreter",
    "https://overpass.kumi.systems/api/interpreter"
]

lat = 40.4168
lon = -3.7038


overpass_query = """
[out:json][timeout:90];
(
  node["amenity"="police"](40.36,-3.75,40.50,-3.60);
  way["amenity"="police"](40.36,-3.75,40.50,-3.60);
  relation["amenity"="police"](40.36,-3.75,40.50,-3.60);
);
out center;
"""

data = None


for url in overpass_urls:
    try:
        print(f"Consultando Overpass API en: {url} ...")

        response = requests.get(url, params={'data': overpass_query}, timeout=100)
        response.raise_for_status()
        data = response.json()
        print("Consulta exitosa.")
        break 
    except Exception as e:
        print(f"Fallo al consultar {url}: {e}")
        time.sleep(2) 


m = folium.Map(location=[lat, lon], zoom_start=12)

if data and 'elements' in data:
    elements = data['elements']
    print(f"Se encontraron {len(elements)} elementos.")
    

    for element in elements:
        lat = element.get('lat') or element.get('center', {}).get('lat')
        lon = element.get('lon') or element.get('center', {}).get('lon')
        name = element.get('tags', {}).get('name', 'Comisaría desconocida')
        
        if lat and lon:
            folium.Marker(
                [lat, lon], 
                popup=name, 
                icon=folium.Icon(color='blue', icon='info-sign')
            ).add_to(m)
else:
    print("No se pudieron obtener datos de ninguna API o la respuesta estaba vacía.")


output_file = "mapa_osm.html"
m.save(output_file)
print(f"Mapa guardado exitosamente en: {os.path.abspath(output_file)}")
print("Si el mapa no se muestra abajo, por favor abra el archivo 'mapa_osm.html' en su navegador web.")
```

## 3. Copernicus API (CDSE): Contexto Espacial

Usaremos el catálogo STAC del Copernicus Data Space Ecosystem para buscar imágenes Sentinel-2.
Nota: La búsqueda de metadatos es abierta, pero la descarga de imágenes requiere autenticación. Aquí visualizaremos los "footprints" (huellas) de las imágenes disponibles.

```python
from pystac_client import Client
import folium
import os

# URL del catálogo STAC de Copernicus Data Space Ecosystem
api_url = "https://catalogue.dataspace.copernicus.eu/stac"

try:
    client = Client.open(api_url)

    # Búsqueda de imágenes Sentinel-2
    # Área de interés (Madrid)
    bbox = [-3.8, 40.3, -3.5, 40.5] 

    # Ampliamos el rango de fechas a algo más reciente y amplio (Enero 2024)
    date_range = "2024-01-01/2024-02-01"

    print(f"Buscando en colección 'SENTINEL-2' para el rango {date_range}...")

    search = client.search(
        collections=["SENTINEL-2"],
        bbox=bbox,
        datetime=date_range,
        max_items=10
    )

    items = search.item_collection()
    print(f"Se encontraron {len(items)} escenas.")

    # Si no encuentra nada, intentamos listar las colecciones disponibles para verificar el nombre
    if len(items) == 0:
        print("No se encontraron imágenes. Verificando colecciones disponibles...")
        collections = client.get_collections()
        print("Colecciones disponibles (ejemplo):")
        for i, col in enumerate(collections):
            print(f"- {col.id}")
            if i >= 5: break

    # Visualizar footprints en el mapa
    m_stac = folium.Map(location=[40.4, -3.65], zoom_start=10)

    # Añadir bounding box de búsqueda
    folium.Rectangle(
        bounds=[[bbox[1], bbox[0]], [bbox[3], bbox[2]]],
        color="red",
        fill=False,
        popup="Área de búsqueda"
    ).add_to(m_stac)

    if len(items) > 0:
        # Añadir footprints de las escenas encontradas
        for item in items:
            # GeoJSON del footprint
            folium.GeoJson(
                item.geometry,
                style_function=lambda x: {'color': 'green', 'fillOpacity': 0.1},
                popup=f"Fecha: {item.datetime}\nID: {item.id}"
            ).add_to(m_stac)

    # Guardar mapa como HTML por si acaso
    output_file = "mapa_copernicus.html"
    m_stac.save(output_file)
    print(f"Mapa guardado en: {os.path.abspath(output_file)}")

    # Devolver mapa
    m_stac

except Exception as e:
    print(f"Error al conectar con el catálogo STAC: {e}")
```

## 4. INSPIRE CSW/WFS: Metadatos y Servicios

Consultaremos un servicio de catálogo (CSW) para buscar datasets INSPIRE.
Ejemplo: Catálogo de la Agencia Europea de Medio Ambiente (EEA).

```python
from owslib.csw import CatalogueServiceWeb

# URL del servicio CSW (Ejemplo: EEA)
csw_url = 'https://sdi.eea.europa.eu/catalogue/srv/eng/csw'

try:
    csw = CatalogueServiceWeb(csw_url)
    print(f"Conectado a: {csw.identification.title}")

    # Buscar datasets relacionados con 'land cover'
    csw.getrecords2(keywords=['land cover'], maxrecords=5, outputschema='http://www.opengis.net/cat/csw/2.0.2')

    print(f"Resultados encontrados: {len(csw.records)}")
    for rec in csw.records:
        print(f"- {csw.records[rec].title}")
        
except Exception as e:
    print(f"Error al conectar o consultar CSW: {e}")
    print("Nota: Los servicios CSW a veces tienen tiempos de inactividad o requieren parámetros específicos.")
```

## 5. NASA VIIRS: Iluminación Nocturna

Visualizaremos la capa de "Earth at Night" (VIIRS) utilizando los servicios de GIBS (Global Imagery Browse Services) de la NASA.
Esto nos permite ver la contaminación lumínica y la extensión urbana.

```python
import folium

# Crear mapa base
m_night = folium.Map(location=[40.4168, -3.7038], zoom_start=6, tiles='CartoDB dark_matter')

# Añadir capa WMS de NASA GIBS (VIIRS Earth at Night 2012)
# Nota: Para datos más recientes (Black Marble), se requiere autenticación en NASA Earthdata.
folium.WmsTileLayer(
    url='https://gibs.earthdata.nasa.gov/wms/epsg4326/best/wms.cgi',
    layers='VIIRS_CityLights_2012',
    fmt='image/png',
    name='NASA VIIRS Earth at Night 2012',
    attr='NASA GIBS',
    transparent=True
).add_to(m_night)

folium.LayerControl().add_to(m_night)

m_night
```
